1. GC Content : the formula computes the fraction of the G or C bases in a given DNA sequence

```python
def GC_content(dna_seq):
  L = len(dna_seq)
  b = 0
  for char in dna_seq:
    if char == "G" or char == "C":
      b+= 1

  return b/L
import numpy as np

def GC_content_np(dna_seq):
  return np.isin(list(dna_seq), ["G", "C"]).mean() #.mean() counts True and divides by total nº of elements in the list
```

```python
#AI generated Strings
print(GC_content("AGCTATCGGCTA")) #0.5
print(GC_content("AAGCCCAATAAACCACTCTGACTGGCCGAATAGGGATATAGGCAACGACATGTGCGGCGACCCTTGCGACAGTGACGCTT"))
print(GC_content_np("AAGCCCAATAAACCACTCTGACTGGCCGAATAGGGATATAGGCAACGACATGTGCGGCGACCCTTGCGACAGTGACGCTT"))
```

2. Hamming distance : for 2 equal-length sequences, see how many positions share the same base

```python
def hamming_distance(seq1, seq2):
  n = 0
  if len(seq1) != len(seq2):
    print("Same-length sequences are necessary")
  for char1, char2 in zip(seq1, seq2):
    if char1 != char2:
      n+= 1
  return n

import numpy as np
def hamming_distance_np(seq1, seq2):
  return (np.array(list(seq1)) != np.array(list(seq2))).sum()
```

```python
a = "GATTACA"
b = "GACTATA"
print(hamming_distance(a, b))
print(hamming_distance_np(a, b))
```

3. Running total: sum all total values in a vector

```python
def running_total(values):
  total = 0
  for v in values:
     total += v
  return total

import numpy as np
def running_total_np(values):
  return np.array(values).sum()
```

```python
values = [4, 5, 6, 7, 8]

print(running_total(values))
print(running_total_np(values))
```

4. Dot product : provides the sum between the product of two complementary elements (same index) in 2 different arrays

```python
def dot_product (value, weights):
  avg = 0
  for v, w in zip(value, weights):
    avg += (v*w)

  return avg

def dot_product_np (value, weights):
  return np.dot(np.array(value), np.array(weights))
```

```python
values = [10, 20, 30, 100]
w = [1, 1, 1, 7]

print(dot_product(values, w))
print(dot_product_np(values, w))
```

5. Geom mean: n-th root of the product of n elements

```python
def geom_mean (values):
  product = 1
  for n in values:
    product *= n

  return (product)**(1/len(values))

import numpy as np
def geom_mean_np (values):
  return np.exp(np.log(values).mean()) #ln(v1) + ln(v2) -> mean of that -> e^(mean) to turn back the scale from the ln(...)

"""
--- Alternative method
def geom_mean_np (values):
  return np.prod(data) ** (1 / len(data))
"""
```

6. Harmonic mean : mean for ratios

```python
def har_mean(values):
  sum = 0
  for v in values:
    sum += 1/v

  return (len(values)/sum)

import numpy as np
def har_mean_np(values):
  return 1/np.mean(1/np.array(values))
```

```python
values = [10, 20, 40, 80]
print(har_mean(values))
print(har_mean_np(values))
```

7. Median : obtain the element in the middle of a list (or the mean of 2 elements in the middle)

```python
def median(values):
  if len(values) % 2 == 0:
    return (values[len(values)//2 - 1] + values[len(values)//2]) / 2
  else: return values[len(values)//2]


import numpy as np
def median_np(values):
  return np.median(values)
```

```python
values_odd = [1, 2, 3]
values_even = [1, 2, 3, 4]

print(median(values_odd))
print(median_np(values_even))
print(median(values_odd))
print(median_np(values_even))
```
